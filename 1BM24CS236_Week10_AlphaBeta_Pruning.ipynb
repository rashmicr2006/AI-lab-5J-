```python
def alphabeta(node, alpha, beta, maximizing):
    if isinstance(node, int):
        return node

    if maximizing:
        value = float('-inf')

        for child in node:
            value = max(value, alphabeta(child, alpha, beta, False))
            alpha = max(alpha, value)

            if alpha >= beta:
                print("Pruned:", node[node.index(child)+1:])
                break

        return value

    else:
        value = float('inf')

        for child in node:
            value = min(value, alphabeta(child, alpha, beta, True))
            beta = min(beta, value)

            if alpha >= beta:
                print("Pruned:", node[node.index(child)+1:])
                break

        return value


# Game tree from the lab question
tree = [
    [[10, 9], [14, 18]],
    [[5, 4], [50, 3]]
]

result = alphabeta(tree, float('-inf'), float('inf'), True)

print("Optimal value:", result)
```
