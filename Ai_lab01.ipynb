```python
def print_board(board):
    print()
    for row in board:
        print(" | ".join(row))
        print("-" * 9)
    print()


def check_winner(board, player):

    for row in board:
        if all(cell == player for cell in row):
            return True


    for col in range(3):
        if all(board[row][col] == player for row in range(3)):
            return True


    if all(board[i][i] == player for i in range(3)):
        return True

    if all(board[i][2 - i] == player for i in range(3)):
        return True

    return False


def board_full(board):
    return all(cell != " " for row in board for cell in row)


def play_game():
    board = [[" " for _ in range(3)] for _ in range(3)]
    current_player = "X"

    print("Tic-Tac-Toe")
    print("Enter row and column numbers from 1 to 3.")

    while True:
        print_board(board)

        try:
            row = int(input(f"Player {current_player}, enter row: ")) - 1
            col = int(input(f"Player {current_player}, enter column: ")) - 1
        except ValueError:
            print("Please enter numbers only.")
            continue

        if row not in range(3) or col not in range(3):
            print("Row and column must be between 1 and 3.")
            continue

        if board[row][col] != " ":
            print("That position is already taken.")
            continue

        board[row][col] = current_player

        if check_winner(board, current_player):
            print_board(board)
            print(f"Player {current_player} wins!")
            break

        if board_full(board):
            print_board(board)
            print("It's a draw!")
            break

        current_player = "O" if current_player == "X" else "X"


if __name__ == "__main__":
    play_game()
```

```python


class VacuumCleaner:
    def __init__(self, location, room_a, room_b):
        self.location = location
        self.rooms = {
            "A": room_a,
            "B": room_b
        }

    def display_state(self):
        print("\nCurrent State:")
        print(f"Vacuum Cleaner Location : Room {self.location}")
        print(f"Room A                 : {self.rooms['A']}")
        print(f"Room B                 : {self.rooms['B']}")

    def clean(self):
        """Clean the current room if it is dirty."""
        if self.rooms[self.location] == "Dirty":
            print(f"Action: SUCK -> Cleaning Room {self.location}")
            self.rooms[self.location] = "Clean"
        else:
            print(f"Room {self.location} is already clean.")

    def move_left(self):
        """Move from B to A."""
        if self.location == "B":
            print("Action: MOVE LEFT -> Room A")
            self.location = "A"
        else:
            print("Vacuum is already in Room A.")

    def move_right(self):
        """Move from A to B."""
        if self.location == "A":
            print("Action: MOVE RIGHT -> Room B")
            self.location = "B"
        else:
            print("Vacuum is already in Room B.")

    def is_goal_reached(self):
        """Check whether both rooms are clean."""
        return (
            self.rooms["A"] == "Clean"
            and self.rooms["B"] == "Clean"
        )

    def run(self):
        """Run the vacuum cleaner agent."""
        print("\n=== Vacuum Cleaner Agent ===")

        self.display_state()


        while not self.is_goal_reached():


            self.clean()


            if self.location == "A":
                self.move_right()
            else:
                self.move_left()

            self.display_state()

        print("\nGoal Reached!")
        print("Both Room A and Room B are CLEAN.")

agent = VacuumCleaner(
    location="A",
    room_a="Dirty",
    room_b="Dirty"
)

agent.run()
```
